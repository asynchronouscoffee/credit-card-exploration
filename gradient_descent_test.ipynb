Gradient Descent is needed to lower the cost function. In this exercise we use hypothetical cost function $y = f(x) = x^2 - 2x.$

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
```

```python
# Creating synthetic data
X_poly = np.linspace(-3,5,81)
print(X_poly[:5], '...', X_poly[-5:])
```

```python
# Define cost function
def cost_function(X):
  return X * (X-2)
y_poly = cost_function(X_poly)
```

```python
# Plot Error Surface of X_poly and y_poly
plt.plot(X_poly, y_poly)
plt.xlabel('Parameter Value')
plt.ylabel('Cost Function')
plt.title('Error Surface')
```

```python
# Creating the custom gradient descent
def gradient(X):
  return (2*X) - 2
x_start = 4.5
learning_rate = 0.75
x_next = x_start - gradient(x_start)*learning_rate
x_next
```

best lr is 0.75

The value above is the next gradient descent step after x = 4.5

```python
# Plotting gradient descent path
plt.plot(X_poly, y_poly)
plt.plot([x_start, x_next],
         [cost_function(x_start), cost_function(x_next)],
         '-o')
plt.xlabel('Parameter Value')
plt.ylabel('Cost Function')
plt.legend(['Error Surface', 'Gradient Descent Path'])
```

The path is correct, it align with the goal of reaching lowest value of Cost Function. But, the algorithm is overstep, need to lower the learning rate variable.

```python
# Giving iteration essence for gradient descent algorithm
iterations = 15
x_path = np.empty(iterations,)
x_path[0] = x_start
for iteration_count in range(1, iterations):
  derivative = gradient(x_path[iteration_count-1])
  x_path[iteration_count] = x_path[iteration_count-1] - (derivative * learning_rate)
x_path
```

We can see that the for loop successfully count the x_path, using current estimate to calculate derivative and find the next estimate.

```python
# Plotting Gradient Descent Path
plt.plot(X_poly, y_poly)
plt.plot(x_path, cost_function(x_path), '-o')
plt.xlabel('Parameter Value')
plt.ylabel('Cost Function')
plt.legend(['Error Survace', 'Gradient Descent Path'])
```
